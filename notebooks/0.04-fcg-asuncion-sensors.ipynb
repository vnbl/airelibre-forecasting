# Identifying Working AireLibre Sensors in Gran Asunción

This notebook is designed to identify functioning sensors in the Gran Asunción area and prepare their raw readings for use in model training. The process follows a set of defined conditions to ensure that only reliable data is used.

## Conditions for Sensor Selection:
- **Location**: The sensors must be located within the Gran Asunción area (Cluster 0). For more details on sensor clustering, refer to the [clustered sensors report](../reports/clustered_sensors.csv).
- **Data Availability**: The sensor must have readings for at least 80% of the analyzed period. The sensors meeting this criterion are listed in the [working sensors report (80%)](../reports/working_sensors_80p).
- **Data Type**: Previous analyses indicate that only particulate matter (PM) readings are available for the sensors. Therefore, only PM readings are considered for this analysis. Other environmental data, such as weather conditions, should be sourced from dedicated weather data sources.

## Key Steps in the Notebook:
1. **Filter Sensors by Location**: Ensure that only sensors within the Gran Asunción area (Cluster 0) are considered.
2. **Check Sensor Data Availability**: Verify that each sensor has data for at least 80% of the period being analyzed.
3. **PM Data Extraction**: Only particulate matter readings (PM2.5 and PM10) are considered for further analysis.
4. **Prepare Data for Model Training**: Raw readings that meet the criteria are cleaned and preprocessed for use in predictive modeling.

## References:
- [Clustered Sensors Report](../reports/clustered_sensors.csv) – Provides sensor locations and their cluster assignments.
- [Working Sensors Report (80%)](../reports/working_sensors_80p) – Lists sensors with at least 80% data availability.


```python
import pandas as pd

# File paths
clustered_sensors_file = '../reports/clustered_sensors.csv'
working_sensors_file = '../reports/working_sensors_80p.csv'
airelibre_data_file = '../data/raw/airelibre_data.csv'
output_file = '../data/interim/filtered_sensor_data.csv'

column_names = [
    "sensor", "source", "description", "version", 
    "pm1dot0", "pm2dot5", "pm10", "humidity", 
    "temperature", "pressure", "co2", 
    "longitude", "latitude", "recorded"
]

try:
    clustered_sensors = pd.read_csv(clustered_sensors_file)
    working_sensors = pd.read_csv(working_sensors_file)
    airelibre_data = pd.read_csv(airelibre_data_file, header=None, names=column_names)
    dtype_mapping = {
        "sensor": "string",
        "source": "string",
        "description": "string",
        "version": "string",
        "pm1dot0": "float64",
        "pm2dot5": "float64",
        "pm10": "float64",
        "humidity": "float64",
        "temperature": "float64",
        "pressure": "float64",
        "co2": "float64",
        "longitude": "float64",
        "latitude": "float64"
    }
    
    airelibre_data = airelibre_data.astype(dtype_mapping)
    
    # Parse the 'recorded' column as datetime
    airelibre_data['recorded'] = pd.to_datetime(
        airelibre_data['recorded'], format='%Y-%m-%dT%H:%M:%S%z', errors='coerce'
    )

    if 'source' not in clustered_sensors.columns or 'cluster' not in clustered_sensors.columns:
        raise KeyError("'clustered_sensors.csv' must contain 'source' and 'cluster' columns.")
    if 'source' not in working_sensors.columns:
        raise KeyError("'working_sensors_80p.csv' must contain a 'source' column.")
    if 'source' not in airelibre_data.columns:
        raise KeyError("'airelibre_data.csv' must contain a 'source' column.")

    # Filter sensors in cluster 0 (Asuncion) and cross reference with working sensors
    cluster_0_sensors = clustered_sensors[clustered_sensors['cluster'] == 0]['source']
    working_sensors_set = set(working_sensors['source'])
    valid_sensors = set(cluster_0_sensors).intersection(working_sensors_set)

    filtered_data = airelibre_data[airelibre_data['source'].isin(valid_sensors)]

    # Save filtered data to a new CSV file
    filtered_data.to_csv(output_file, index=False)
    print(f"Filtered sensor data saved to {output_file}")

except FileNotFoundError as e:
    print(f"Error: {e}")
except KeyError as e:
    print(f"Error: {e}")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```

## Data Cleaning

The data cleaning process prepares the PM measurements for analysis by focusing on the following steps:

1. **Load Data**: The dataset is loaded with columns for sensor ID (`source`), PM measurements (`pm2dot5`, `pm1dot0`, `pm10`), and timestamps (`recorded`).

2. **Convert Timestamps**: The `recorded` column is converted to datetime format, and rows with invalid timestamps are removed.

3. **Set Index**: The `recorded` column is set as the index to facilitate time-based operations.

4. **Select Numeric Columns**: Only numeric columns (PM readings) are selected for further processing.

5. **Resample Data**: The data is resampled to hourly intervals, calculating the average PM values for each hour per sensor.

6. **Outlier Handling**: Outlier detection is skipped for PM readings, as these values naturally fluctuate in air quality monitoring.


```python
import pandas as pd
import plotly.express as px

# File path to the filtered data
filtered_data_file = '../data/interim/filtered_sensor_data.csv'

try:

    filtered_data = pd.read_csv(filtered_data_file, usecols=['source', 'pm2dot5', 'pm1dot0','pm10','recorded'])

    filtered_data['recorded'] = pd.to_datetime(
        filtered_data['recorded'], format='%Y-%m-%d %H:%M:%S%z', errors='coerce'
    )

    filtered_data = filtered_data.dropna(subset=['recorded'])

    filtered_data = filtered_data.set_index('recorded')

    numeric_columns = filtered_data.select_dtypes(include=['number']).columns

    filtered_data_resampled = (
        filtered_data
        .groupby('source')  # Keep 'source' in the result
        .apply(lambda group: group.resample('h').mean()[numeric_columns])#.interpolate())
    )

    filtered_data_resampled = filtered_data_resampled.reset_index()

    print(filtered_data_resampled.head())

    fig = px.line(
        filtered_data_resampled,  
        x='recorded', 
        y='pm2dot5', 
        title='PM2.5 Levels Over Time',
        labels={
            'recorded': 'Time (Recorded)',
            'pm2dot5': 'PM2.5 Levels'
        },
        template='plotly',
        color='source'  # Group data by 'source' and plot separate lines for each group
    )   

    fig.update_layout(
        xaxis_title='Time (Recorded)',
        yaxis_title='PM2.5 Levels',
        xaxis=dict(showgrid=True, title_font=dict(size=14)),
        yaxis=dict(showgrid=True, title_font=dict(size=14)),
        legend_title_text='PM2.5',
    )

    # Show the plot
    fig.show()

except FileNotFoundError as e:
    print(f"Error: {e}")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```

## Handling Missing Data with KNN Imputation

### Overview
The dataset contains significant missing values, which could impact the quality of analysis. To address this, we used **K-Nearest Neighbors (KNN) Imputation**, a method that predicts missing values based on similarities with other data points.

### Steps

1. **Reshape the Data**:
   - The data was transformed into a long format with columns for `recorded` (timestamps), `source` (sensor ID), and `value` (PM readings).
   - This format simplifies handling and ensures compatibility with imputation.

2. **Pivot the Data**:
   - The dataset was pivoted to create a wide format where each sensor and measurement type (`pm2dot5`, `pm1dot0`, `pm10`) became separate columns.
   - This structure aligns sensor readings horizontally for easier imputation.

3. **Apply KNN Imputation**:
   - The **KNNImputer** from `scikit-learn` was applied with 15 neighbors (`n_neighbors=15`).
   - Missing values were filled using the average of the nearest 15 rows based on Euclidean distance.

4. **Revert to Original Format**:
   - The imputed data was transformed back into its original long format for consistency.
   - Columns for `source`, `recorded`, and PM measurements were restored.

5. **Save and Visualize Results**:
   - The cleaned dataset was saved for future use.
   - A line plot was generated to visualize PM2.5 levels over time, grouped by sensor.

### Why KNN Imputation?
- KNN leverages similarities between nearby data points to fill missing values.
- This method works well for air quality data, where measurements are often spatially and temporally correlated.


```python
from sklearn.impute import KNNImputer
import pandas as pd

filtered_data_resampled['recorded'] = pd.to_datetime(filtered_data_resampled['recorded'])

# melt dataframe for easier handling
if 'value' not in filtered_data_resampled.columns:
    filtered_data_resampled = filtered_data_resampled.melt(
        id_vars=['recorded', 'source'],
        value_vars=['pm2dot5', 'pm1dot0', 'pm10'],
        var_name='variable',
        value_name='value'
    )

imputed_data = filtered_data_resampled.set_index('recorded').sort_index()

# pivot the data to align sensors horizontally (one row per timestamp per variable) and flatten
pivoted_data = imputed_data.pivot_table(index=imputed_data.index, columns=['source', 'variable'], values='value')
pivoted_data.columns = [f"{source}_{variable}" for source, variable in pivoted_data.columns]

# KNN imputation
knn_imputer = KNNImputer(n_neighbors=15)
imputed_array = knn_imputer.fit_transform(pivoted_data)

# convert dataframe to original format and unpivot
imputed_pivoted = pd.DataFrame(imputed_array, index=pivoted_data.index, columns=pivoted_data.columns)
imputed_data = (
    imputed_pivoted.stack()
    .reset_index()
    .rename(columns={'level_0': 'recorded', 'level_1': 'source_variable', 0: 'value'})
)

imputed_data[['source', 'variable']] = imputed_data['source_variable'].str.rsplit('_', n=1, expand=True)

imputed_data = imputed_data.pivot(index=['recorded', 'source'], columns='variable', values='value').reset_index()

imputed_data['recorded'] = pd.to_datetime(imputed_data['recorded'])
numeric_columns = ['pm2dot5', 'pm1dot0', 'pm10']
for col in numeric_columns:
    if col in imputed_data.columns:
        imputed_data[col] = imputed_data[col].astype(float)
    else:
        print(f"Warning: {col} not found in DataFrame.")

print(imputed_data.head())

# Save and display results

imputed_data.to_csv('../data/interim/filtered_pm_asuncion_resampled.csv', index=False)

fig = px.line(
    imputed_data, 
    x='recorded', 
    y='pm2dot5', 
    title='PM2.5 Levels Over Time',
    labels={
        'recorded': 'Time (Recorded)',
        'pm2dot5': 'PM2.5 Levels'
    },
    template='plotly',
    color='source'  # Group data by 'source' and plot separate lines for each group
)   

# Update layout for better readability
fig.update_layout(
    xaxis_title='Time (Recorded)',
    yaxis_title='PM2.5 Levels',
    xaxis=dict(showgrid=True, title_font=dict(size=14)),
    yaxis=dict(showgrid=True, title_font=dict(size=14)),
    legend_title_text='PM2.5',
)

# Show the plot
fig.show()
```
